```python
def save_plot_as_html(fig, filename, insight):
    global plot_containers
    if 'plot_containers' not in globals():
        plot_containers = ""

    file_path = (html_file_path / filename).resolve()
    file_path.parent.mkdir(parents=True, exist_ok=True)

    html_content = pio.to_html(fig, full_html=False, include_plotlyjs='inline')
    plot_containers += f'''
    <div class="plot-container" id="{filename}" onclick="openPlot('{filename}')">
       <div class="plot">{html_content}</div>
       <div class="insight">{insight}</div>
    </div>
    '''

    fig.write_html(str(file_path), full_html=False, include_plotlyjs='inline')
    print(f"Saved plot to: {file_path}")
    print(f"File exists after save: {file_path.exists()}")
```

```python
import pandas as pd

apps = pd.read_csv("apps.csv")
reviews = pd.read_csv("user_reviews.csv")

print("Apps dataset shape:", apps.shape)
print("User Reviews dataset shape:", reviews.shape)
```

```python
print("APPS DATASET")
print(apps.info())

print("\nUSER REVIEWS DATASET")
print(reviews.info())
```

```python
# ============================================================
# TASK 1 - APP SIZE vs RATING BUBBLE ANALYSIS
# ============================================================

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

# Required categories
task1_categories = [
    'GAME',
    'BEAUTY',
    'BUSINESS',
    'COMICS',
    'COMMUNICATION',
    'DATING',
    'ENTERTAINMENT',
    'SOCIAL',
    'EVENTS'
]

# ------------------------------------------------------------
# FILTER DATA
# ------------------------------------------------------------

# Build the missing rating dataset from the existing apps + reviews tables
task1_rating_filtered = apps.copy()

# Convert installs to numeric
task1_rating_filtered["Installs_Numeric"] = (
    task1_rating_filtered["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
    .str.replace(" ", "", regex=False)
    .replace("", pd.NA)
    .astype(float)
)

# Ensure category names are comparable with task1_categories
task1_rating_filtered["Category"] = (
    task1_rating_filtered["Category"]
    .astype(str)
    .str.upper()
    .str.strip()
)

# Aggregate review subjectivity per app
review_subjectivity = (
    reviews.dropna(subset=["App", "Sentiment_Subjectivity"])
    .groupby("App", as_index=False)["Sentiment_Subjectivity"]
    .mean()
    .rename(columns={"Sentiment_Subjectivity": "Avg_Sentiment_Subjectivity"})
)

# Merge the sentiment summary back to app data
task1_rating_filtered = task1_rating_filtered.merge(
    review_subjectivity,
    on="App",
    how="left"
)

# Filter to the task-specific subset
task1_rating_filtered = task1_rating_filtered[
    (task1_rating_filtered["Category"].isin(task1_categories)) &
    (task1_rating_filtered["Reviews"] > 500) &
    (task1_rating_filtered["Installs_Numeric"] > 50000) &
    (task1_rating_filtered["Avg_Sentiment_Subjectivity"] > 0.5) &
    (~task1_rating_filtered["App"].str.contains("S", case=False, na=False))
].copy()

# Keep the downstream dataframe name expected by the rest of the cell
task1_df = task1_rating_filtered.copy()

# ------------------------------------------------------------
# CATEGORY DISPLAY TRANSLATIONS
# ------------------------------------------------------------

task1_df['Category_Display'] = task1_df['Category'].replace({
    'BEAUTY': 'सौंदर्य',
    'BUSINESS': 'வணிகம்',
    'DATING': 'Dating'
})

# ------------------------------------------------------------
# BUBBLE CHART
# ------------------------------------------------------------

fig_task1 = px.scatter(
    task1_df,
    x='Size',
    y='Rating',
    size='Installs_Numeric',
    color='Category_Display',
    hover_name='App',
    hover_data={
        'Size': ':.1f',
        'Rating': ':.1f',
        'Reviews': ':,',
        'Installs_Numeric': ':,',
        'Avg_Sentiment_Subjectivity': ':.3f',
        'Category_Display': True
    },
    labels={
        'Size': 'App Size (MB)',
        'Rating': 'Average Rating',
        'Installs_Numeric': 'Installs',
        'Category_Display': 'Category'
    },
    title='App Size vs Average Rating'
)

# ------------------------------------------------------------
# HIGHLIGHT GAME
# ------------------------------------------------------------

for trace in fig_task1.data:
    if trace.name == 'GAME':
        trace.marker.color = 'pink'
        trace.marker.line = dict(
            color='white',
            width=1
        )

# ------------------------------------------------------------
# DASHBOARD STYLE
# ------------------------------------------------------------

fig_task1.update_layout(
    plot_bgcolor='black',
    paper_bgcolor='black',
    font_color='white',

    title_font={
        'size': 18
    },

    xaxis=dict(
        title='App Size (MB)',
        title_font={'size': 12},
        gridcolor='#333333',
        zeroline=False
    ),

    yaxis=dict(
        title='Average Rating',
        title_font={'size': 12},
        range=[3.5, 5],
        gridcolor='#333333',
        zeroline=False
    ),

    legend=dict(
        title='Category',
        font={'size': 10}
    ),

    hoverlabel=dict(
        bgcolor='#111111',
        font_color='white'
    ),

    margin=dict(
        l=20,
        r=20,
        t=50,
        b=20
    ),

    width=900,
    height=550
)

fig_task1.update_traces(
    marker=dict(
        opacity=0.75
    )
)

# ------------------------------------------------------------
# SAVE INTO YOUR EXISTING DASHBOARD SYSTEM
# ------------------------------------------------------------

save_plot_as_html(
    fig_task1,
    "Task 1 - App Size Rating Bubble.html",
    "Interactive analysis showing the relationship between app size and average rating for qualified applications. Bubble size represents installs, GAME is highlighted in pink, and Beauty, Business and Dating use the required display labels."
)

print("Task 1 graph generated successfully.")
print("Qualified applications:", len(task1_df))
```

```python
#Data Preparation

task6_df = apps.copy()

task6_df["Rating"] = pd.to_numeric(task6_df["Rating"], errors="coerce")
task6_df["Reviews"] = pd.to_numeric(task6_df["Reviews"], errors="coerce")
task6_df["Size"] = pd.to_numeric(task6_df["Size"], errors="coerce")

task6_df["Installs_Numeric"] = pd.to_numeric(
    task6_df["Installs"].astype(str).str.replace("[+,]", "", regex=True),
    errors="coerce"
)

task6_df["Price_Numeric"] = pd.to_numeric(
    task6_df["Price"].astype(str).str.replace("$", "", regex=False),
    errors="coerce"
)

task6_df["Android_Version"] = pd.to_numeric(
    task6_df["Android Ver"].astype(str).str.extract(r"(\d+(?:\.\d+)?)")[0],
    errors="coerce"
)

task6_df["Revenue"] = (
    task6_df["Installs_Numeric"] * task6_df["Price_Numeric"]
)

task6_df["App_Name_Length"] = task6_df["App"].astype(str).str.len()

task6_df = task6_df.dropna(
    subset=[
        "App",
        "Category",
        "Installs_Numeric",
        "Price_Numeric",
        "Size",
        "Android_Version",
        "Content Rating"
    ]
).copy()

task6_df = task6_df[
    (task6_df["Installs_Numeric"] >= 10000) &
    (task6_df["Android_Version"] > 4.0) &
    (task6_df["Size"] > 15) &
    (task6_df["Content Rating"] == "Everyone") &
    (task6_df["App_Name_Length"] <= 30)
].copy()

top3_categories = (
    task6_df
    .groupby("Category")["Installs_Numeric"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
    .index
)

task6_df = task6_df[
    task6_df["Category"].isin(top3_categories)
].copy()

task6_df[[
    "App",
    "Category",
    "Type",
    "Installs_Numeric",
    "Price_Numeric",
    "Revenue",
    "Size",
    "Android_Version",
    "Content Rating",
    "App_Name_Length"
]].head()
```

```python
#Aggregation and Analysis

task6_analysis = task6_df.copy()

task6_analysis["Revenue_Eligible"] = (
    task6_analysis["Revenue"] >= 10000
)

task6_summary = (
    task6_analysis
    .groupby(["Category", "Type"], as_index=False)
    .agg(
        Average_Installs=("Installs_Numeric", "mean"),
        Average_Revenue=("Revenue", "mean"),
        Total_Installs=("Installs_Numeric", "sum"),
        Total_Revenue=("Revenue", "sum"),
        App_Count=("App", "count")
    )
)

top3_categories = (
    task6_analysis
    .groupby("Category")["Installs_Numeric"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
    .index
)

task6_summary = task6_summary[
    task6_summary["Category"].isin(top3_categories)
].copy()

task6_summary = task6_summary.sort_values(
    ["Category", "Type"]
)

task6_summary
```

```python
#Final Dual-Axis Visualization

fig_task6 = go.Figure()

fig_task6.add_trace(
    go.Bar(
        x=[f"{cat} - {typ}" for cat, typ in zip(
            task6_summary["Category"],
            task6_summary["Type"]
        )],
        y=task6_summary["Average_Installs"],
        name="Average Installs",
        yaxis="y",
        hovertemplate=(
            "<b>%{x}</b><br>"
            "Average Installs: %{y:,.0f}"
            "<extra></extra>"
        )
    )
)

fig_task6.add_trace(
    go.Scatter(
        x=[f"{cat} - {typ}" for cat, typ in zip(
            task6_summary["Category"],
            task6_summary["Type"]
        )],
        y=task6_summary["Average_Revenue"],
        name="Average Revenue",
        mode="lines+markers",
        yaxis="y2",
        hovertemplate=(
            "<b>%{x}</b><br>"
            "Average Revenue: $%{y:,.2f}"
            "<extra></extra>"
        )
    )
)

fig_task6.update_layout(
    title="Average Installs vs Revenue: Free and Paid Apps",
    plot_bgcolor="black",
    paper_bgcolor="black",
    font_color="white",
    title_font={"size": 18},
    xaxis=dict(
        title="Category and App Type",
        tickangle=-45
    ),
    yaxis=dict(
        title="Average Installs",
        gridcolor="#333333",
        zeroline=False
    ),
    yaxis2=dict(
        title="Average Revenue ($)",
        overlaying="y",
        side="right",
        showgrid=False
    ),
    legend=dict(
        title="Metric"
    ),
    hoverlabel=dict(
        bgcolor="#111111",
        font_color="white"
    ),
    margin=dict(l=20, r=70, t=50, b=110),
    width=900,
    height=550
)

save_plot_as_html(
    fig_task6,
    "Task 6 - Average Installs Revenue.html",
    "Interactive dual-axis visualization comparing average installs and average revenue between Free and Paid apps across the top three app categories. The analysis applies the required install, revenue, Android version, size, content rating, and app name length criteria. The visualization is intended for display between 1 PM and 2 PM IST in the final master dashboard."
)

display(fig_task6)
```
